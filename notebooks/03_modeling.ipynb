```python
from pathlib import Path

FEATURE_PATH = (
    Path("..")
    / "data"
    / "processed"
    / "window_features.parquet"
)
```

```python
feature_df = pd.read_parquet(FEATURE_PATH)

print("Feature table loaded successfully.")
print("Rows:", len(feature_df))
print("Columns:", len(feature_df.columns))
```

```python
participants = (
    feature_df["participant_id"]
    .drop_duplicates()
    .tolist()
)

print("Number of participants:", len(participants))
print("Participants:", participants)
```

```python
train_participants, test_participants = train_test_split(
    participants,
    test_size=0.2,
    random_state=42
)

print("Training participants:", train_participants)
print("Testing participants:", test_participants)
```

```python
train_df = feature_df[
    feature_df["participant_id"].isin(train_participants)
].copy()

test_df = feature_df[
    feature_df["participant_id"].isin(test_participants)
].copy()

print("Training windows:", len(train_df))
print("Testing windows:", len(test_df))
```

```python
common_participants = set(train_participants).intersection(
    set(test_participants)
)

print("Participants in both sets:", common_participants)
```
